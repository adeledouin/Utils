```python
from Utils.CleanPlot.classPlot import *
logging.basicConfig(format='%(asctime)s - %(funcName)s() | %(message)s', level=logging.INFO)

plot = PaperPlot(latex=True)

x = np.arange(-100, 100)

def gaussian_noise(z, mu, sigma):
    return 1/(sigma * np.sqrt(2*np.pi)) * np.exp(-(z-mu)**2/(2*sigma**2))

fig, ax = plot.belleFigure('$x$', '$\phi (x)$', nfigure=None)
ax.plot(x, gaussian_noise(x, mu=0, sigma=1),  '.',  color='C0', label='$\mu = {} | \sigma = {}$'.format(0, 1))
ax.plot(x, gaussian_noise(x, mu=10, sigma=10),  '.',  color='C1', label='$\mu = {} | \sigma = {}$'.format(10, 10))
plot.plt.grid(True, which='both')
plot.plt.legend(loc="best")
save = None
plot.fioritures(ax, fig, title=None, label=True, grid=None, save=save)
```
